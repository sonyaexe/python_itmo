```python
import pandas as pd
```

```python
df = pd.read_csv('train.csv')
df
```

```python
df.info()
```

```python
pclass_total = df.groupby("Pclass")["PassengerId"].count()
survival_rate = pd.DataFrame({"Survival": df.groupby("Pclass")["Survived"].sum() / pclass_total * 100})
survival_rate
```

```python
df['PopularName'] = df['Name'].str.split(',').str[0]
popular_names = df.groupby('Sex')['PopularName'].value_counts()
print(popular_names.groupby(level=0).head(1))
```

```python
popular_names_plass = df.groupby(['Pclass', 'Sex'])['PopularName'].value_counts()
print(popular_names_plass.groupby(level=[0, 1]).head(1))
```
