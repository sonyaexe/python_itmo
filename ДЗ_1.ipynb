```python
class Chocolate:
  def __init__(self, price: float, additive: str = None):

    if price > 0:
      self.price = price
    else:
      raise ValueError('Какая-то странная цена... Цена не может быть 0 рублей или меньше!')

    self.additive = additive

  def show_my_chocolate(self):
    if self.additive is not None:
      print(f'Шоколадный батончик, добавка: {self.additive}')

    else:
      print('Обычный шоколадный батончик')


  def __str__(self):
    return f'Шоколадный батончик, {self.price} рублей'


chocolate = Chocolate(20, 'карамель, нуга')
print(chocolate)


chocolate.show_my_chocolate()


```

```python
chocolate = Chocolate(0, 'миндаль')
print(chocolate)
```

```python
chocolate = Chocolate(50.2)
print(chocolate)
```

```python
chocolate = Chocolate(None, 333)
print(chocolate)
```

```python
chocolate = Chocolate(100, 333)
print(chocolate)
```
